```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Task 1: Load CSV, Display Data, Missing Values & Summary Statistics

```python
# Load dataset
df = pd.read_csv("data.csv")

# Display first 5 rows
df.head()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Summary statistics of numerical columns
df.describe()
```

### Task 2: Handle Missing Values (Fill with Median)

```python
# Numerical columns
num_cols = df.select_dtypes(include=np.number).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())

# Categorical columns
cat_cols = df.select_dtypes(include="object").columns
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

# Verify
df.isnull().sum()
```

### Task 3: Filter Rows Based on Threshold

```python
filtered_df = df[df["Index"] > 50]
filtered_df.head()
```

```python
# Example
df[df["Index"].between(20, 40)]
```

### 🔹 Task 4: Group By Categorical Column

```python
# Example: Group by Country
country_stats = df.groupby("Country")["Index"].agg(["mean", "std", "count"])
country_stats.head()
```

# 📊 Visualization: Customers per Country (Top 10)

```python
plt.figure(figsize=(10,5))
df["Country"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Countries by Number of Customers")
plt.xlabel("Country")
plt.ylabel("Customer Count")
plt.xticks(rotation=45)
plt.show()
```

### 🔹 Task 5: Merge Two DataFrames

```python
# Create Second DataFrame (Example: Customer Type)
df_extra = pd.DataFrame({
    "Customer Id": df["Customer Id"].head(10),
    "Customer_Type": ["Premium", "Regular", "Premium", "Regular", "Premium",
                      "Regular", "Premium", "Regular", "Premium", "Regular"]
})
```

```python
merged_df = pd.merge(df, df_extra, on="Customer Id", how="left")
merged_df.head()
```

## 🔹 Task 6: Create New Column

```python
# Example: Full Name
df["Full Name"] = df["First Name"] + " " + df["Last Name"]
df.head()
```

```python
df["Company_Country"] = df["Company"] + " (" + df["Country"] + ")"
```

#🔹 Task 7: Correlation Matrix

```python
# Your dataset has limited numeric data, so correlation is mainly on Index.
corr = df[num_cols].corr()
corr
```

```python
# 📊 Heatmap
plt.figure(figsize=(5,4))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

# 🔹 Task 8: Scatter Plot

```python
country_count = df["Country"].value_counts().reset_index()
country_count.columns = ["Country", "Customer_Count"]

plt.figure(figsize=(8,5))
plt.scatter(country_count.index, country_count["Customer_Count"])
plt.xlabel("Country Index")
plt.ylabel("Number of Customers")
plt.title("Scatter Plot: Country vs Customer Count")
plt.grid(True)
plt.show()
```

### 📦 Pie Chart: Customer Distribution

```python
df["Country"].value_counts().head(5).plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6,6)
)
plt.title("Top 5 Countries Customer Share")
plt.ylabel("")
plt.show()
```

### 📦 City-wise Customer Count

```python
plt.figure(figsize=(10,5))
df["City"].value_counts().head(10).plot(kind="barh")
plt.title("Top 10 Cities by Customers")
plt.xlabel("Count")
plt.show()
```
