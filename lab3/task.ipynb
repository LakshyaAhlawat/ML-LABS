## Lab Task 3: Performance Measures (Without Any Model)

### Objective:
To calculate evaluation metrics using given Actual and Predicted outputs.

### Given:
- Actual labels (Y): 1, 0, 1, 0, 1, 0, 1, 0
- Predicted labels (Ŷ): 1, 1, 1, 0, 0, 0, 1, 0


### STEP-1
** TP(True Positive) -> Actual=1 , Predicted=1
** TN(True Negative)-> Actual=0, Predicted=0;
** FP(False Positive)-> Actual=0 , Predicted=1
** FN(False Negative)-> Actual=1 , Predicted=0

```python
# actual output labels
y_true=[1,0,1,0,1,0,1,0]

# Predicted output labels
y_pred = [1, 1, 1, 0, 0, 0, 1, 0]
```

### Initialise TP,TN,FP,FN

```python
TP=0
TN=0
FP=0
FN=0
```

Count TP, TN, FP, FN

```python
for actual,predicted in zip(y_true,y_pred):
  if actual==1 and predicted==1:
    TP+=1
  elif actual==0 and predicted==0:
    TN+=1
  elif actual==1 and predicted==0:
    FN+=1
  else:
    FP+=1
```

### Display TP, TN, FP, FN

```python
print("True Positives (TP):", TP)
print("True Negatives (TN):", TN)
print("False Positives (FP):", FP)
print("False Negatives (FN):", FN)
```

## Confusion Matrix

```python
accuracy=(TP+TN)/(TP+TN+FP+FN)
accuracy
```

# Calculate Precision

```python
precision = TP / (TP + FP)
precision
# Out of all predicted positives, how many are correct?
```

# Calculate Recall (Sensitivity)

```python
recall = TP / (TP + FN)
recall
# Out of all actual positives, how many did we correctly detect?
```

## Calculate F1-Score

```python
f1_score = 2 * (precision * recall) / (precision + recall)
f1_score
# F1 score balances Precision & Recall

# Useful when dataset is imbalanced
```

## Calculate Specificity

```python
specificity = TN / (TN + FP)
specificity
# Measures how well negative cases are identified
```

## Final Results

### Final Evaluation Metrics

- **Accuracy:** 0.75  
- **Precision:** 0.75  
- **Recall:** 0.75  
- **F1-Score:** 0.75  
- **Specificity:** 0.75  

# Real-Life Example

### Case where Recall is more important than Precision

**Medical Diagnosis (Cancer / COVID Detection)**

- Missing a sick patient (False Negative) can be life-threatening
- Extra false alarms are acceptable
- Hence, **Recall is more important than Precision**
