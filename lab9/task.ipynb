### Cell 1: Imports

```python
import heapq   # for priority queue (used in Greedy & A*)
```

### Cell 2: Graph Definition (Adjacency List)

```python
# Graph: node -> list of (neighbor, cost)
graph = {
    'S': [('A', 1), ('B', 4)],
    'A': [('C', 2), ('D', 5)],
    'B': [('D', 1)],
    'C': [('G', 5)],
    'D': [('G', 1)],
    'G': []
}
```

### Cell 3: Heuristic Values h(n)

```python
heuristic = {
    'S': 7,
    'A': 6,
    'B': 2,
    'C': 1,
    'D': 1,
    'G': 0
}
```

# STEP 1: DFS (Depth First Search)

### Cell 4: DFS Implementation

```python
def dfs(graph, start, goal):
    stack = [(start, [start])]
    visited = set()
    
    while stack:
        node, path = stack.pop()
        
        if node not in visited:
            visited.add(node)
            
            if node == goal:
                return path
            
            for neighbor, _ in reversed(graph[node]):
                stack.append((neighbor, path + [neighbor]))
    
    return None
```

### Cell 5: Run DFS

```python
dfs_path = dfs(graph, 'S', 'G')
print("DFS Path:", dfs_path)
```

# STEP 2: Greedy Best-First Search

### Cell 6: Greedy BFS Implementation

```python
def greedy_bfs(graph, heuristic, start, goal):
    pq = [(heuristic[start], start, [start])]
    visited = set()
    
    while pq:
        h, node, path = heapq.heappop(pq)
        
        if node in visited:
            continue
        
        visited.add(node)
        
        if node == goal:
            return path, visited
        
        for neighbor, _ in graph[node]:
            heapq.heappush(pq, (heuristic[neighbor], neighbor, path + [neighbor]))
    
    return None, visited
```

### Cell 7: Run Greedy

```python
greedy_path, greedy_visited = greedy_bfs(graph, heuristic, 'S', 'G')

print("Greedy Path:", greedy_path)
print("Nodes Explored:", greedy_visited)
```

# STEP 3: A* Search

### Cell 8: A* Implementation

```python
def a_star(graph, heuristic, start, goal):
    pq = [(0 + heuristic[start], 0, start, [start])]  # (f, g, node, path)
    visited = set()
    
    while pq:
        f, g, node, path = heapq.heappop(pq)
        
        if node in visited:
            continue
        
        visited.add(node)
        
        if node == goal:
            return path, g, visited
        
        for neighbor, cost in graph[node]:
            new_g = g + cost
            new_f = new_g + heuristic[neighbor]
            heapq.heappush(pq, (new_f, new_g, neighbor, path + [neighbor]))
    
    return None, None, visited
```

### Cell 9: Run A*

```python
a_star_path, a_star_cost, a_star_visited = a_star(graph, heuristic, 'S', 'G')

print("A* Path:", a_star_path)
print("Total Cost:", a_star_cost)
print("Nodes Explored:", a_star_visited)
```

# STEP 4: Path Cost Function

### Cell 10:

```python
def path_cost(graph, path):
    cost = 0
    for i in range(len(path)-1):
        for neighbor, c in graph[path[i]]:
            if neighbor == path[i+1]:
                cost += c
    return cost
```

### Cell 11:

```python
dfs_cost = path_cost(graph, dfs_path)
greedy_cost = path_cost(graph, greedy_path)

print("DFS Cost:", dfs_cost)
print("Greedy Cost:", greedy_cost)
print("A* Cost:", a_star_cost)
```
